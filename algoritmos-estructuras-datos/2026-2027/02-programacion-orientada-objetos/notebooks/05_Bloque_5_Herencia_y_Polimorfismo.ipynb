# 5.1. Herencia: superclase, subclases e instancias

## Superclase

`Producto` contendrá los datos comunes a todos los productos:

- referencia;
- nombre;
- PVP;
- descripción.

Las clases más específicas heredarán estos elementos.

La idea es evitar repetir código innecesariamente.

````python
class Producto:

    def __init__(self, referencia, nombre, pvp, descripcion):
        # Atributos comunes a cualquier producto.
        self.referencia = referencia
        self.nombre = nombre
        self.pvp = pvp
        self.descripcion = descripcion

    def __str__(self):
        # Representación común de un producto.
        return (
            f"REFERENCIA: {self.referencia}\n"
            f"NOMBRE: {self.nombre}\n"
            f"PVP: {self.pvp:.2f} €\n"
            f"DESCRIPCIÓN: {self.descripcion}"
        )
````

## Subclases

Una subclase se declara indicando entre paréntesis la clase de la que hereda:

```python
class Alimento(Producto):
```

Eso significa:

> «Alimento es un tipo especializado de Producto».

La subclase hereda los atributos y métodos de la clase base y, además, puede:

- añadir nuevos atributos;
- añadir nuevos métodos;
- redefinir métodos heredados.

## Utilizar `super()`

Cuando una subclase necesita inicializar también la parte heredada, podemos llamar al constructor de la clase base mediante:

```python
super().__init__(...)
```

Esto evita duplicar la lógica de `Producto`.

````python
class Alimento(Producto):

    def __init__(
        self,
        referencia,
        nombre,
        pvp,
        descripcion,
        productor,
        distribuidor
    ):
        # Inicializamos la parte heredada de Producto.
        super().__init__(
            referencia,
            nombre,
            pvp,
            descripcion
        )

        # Atributos específicos de Alimento.
        self.productor = productor
        self.distribuidor = distribuidor

    def __str__(self):
        # Reutilizamos primero el texto generado por Producto.
        texto_producto = super().__str__()

        # Después añadimos la información específica.
        return (
            f"{texto_producto}\n"
            f"PRODUCTOR: {self.productor}\n"
            f"DISTRIBUIDOR: {self.distribuidor}"
        )


class Libro(Producto):

    def __init__(
        self,
        referencia,
        nombre,
        pvp,
        descripcion,
        isbn,
        autor
    ):
        # Inicializamos los atributos heredados.
        super().__init__(
            referencia,
            nombre,
            pvp,
            descripcion
        )

        # Atributos específicos de Libro.
        self.isbn = isbn
        self.autor = autor

    def __str__(self):
        texto_producto = super().__str__()

        return (
            f"{texto_producto}\n"
            f"ISBN: {self.isbn}\n"
            f"AUTOR: {self.autor}"
        )


class Adorno(Producto):
    # pass indica que no añadimos nada nuevo.
    #
    # Adorno heredará:
    # - __init__
    # - __str__
    # de Producto.
    pass
````

## Crear instancias de las subclases

Cada subclase puede recibir tanto los datos heredados como sus datos específicos.

````python
alimento = Alimento(
    referencia=2035,
    nombre="Botella de Aceite de Oliva",
    pvp=5,
    descripcion="250 ML",
    productor="La Aceitera",
    distribuidor="Distribuciones SA"
)

libro = Libro(
    referencia=2036,
    nombre="Cocina Mediterránea",
    pvp=9,
    descripcion="Recetas sanas y buenas",
    isbn="0-123456-78-9",
    autor="Doña Juana"
)

adorno = Adorno(
    referencia=2034,
    nombre="Vaso adornado",
    pvp=15,
    descripcion="Vaso de porcelana"
)

print(alimento)
print()
print(libro)
print()
print(adorno)
````

## Comprobar la relación de herencia

`isinstance()` permite comprobar si un objeto es instancia de una clase.

También tiene en cuenta la herencia.

````python
print("¿alimento es Alimento?", isinstance(alimento, Alimento))
print("¿alimento es Producto?", isinstance(alimento, Producto))

print("¿libro es Libro?", isinstance(libro, Libro))
print("¿libro es Producto?", isinstance(libro, Producto))

print("¿adorno es Adorno?", isinstance(adorno, Adorno))
print("¿adorno es Producto?", isinstance(adorno, Producto))
````

También podemos preguntar si una clase hereda de otra con `issubclass()`.

````python
print("¿Alimento hereda de Producto?", issubclass(Alimento, Producto))
print("¿Libro hereda de Producto?", issubclass(Libro, Producto))
print("¿Adorno hereda de Producto?", issubclass(Adorno, Producto))
````

# 5.2. Polimorfismo

El **polimorfismo** permite trabajar de manera uniforme con objetos de tipos diferentes siempre que respondan a las operaciones que necesitamos.

En nuestro ejemplo:

- `Alimento` tiene su propio `__str__`;
- `Libro` tiene su propio `__str__`;
- `Adorno` utiliza el `__str__` heredado de `Producto`.

Podemos almacenar todos esos objetos en una misma lista.

````python
# Una misma estructura de datos contiene objetos de diferentes clases.
productos = [
    adorno,
    alimento,
    libro
]

# La variable producto irá referenciando objetos de distintos tipos.
for producto in productos:

    # El código es el mismo en todas las iteraciones.
    #
    # Sin embargo, Python ejecutará el __str__ correspondiente
    # al tipo REAL del objeto.
    print(producto)

    print("-" * 50)
````

### ¿Qué ocurre en el bucle?

Cuando `producto` referencia a un `Adorno`:

```python
print(producto)
```

termina utilizando el `__str__` heredado de `Producto`.

Cuando referencia a un `Alimento`, utiliza:

```python
Alimento.__str__
```

Cuando referencia a un `Libro`, utiliza:

```python
Libro.__str__
```

El código del bucle no cambia.

Ese comportamiento es un ejemplo de **polimorfismo**.

## Otro ejemplo sencillo de polimorfismo

No tiene por qué aparecer únicamente mediante `__str__`.

También puede darse con métodos normales.

````python
class Animal:

    def hablar(self):
        print("Sonido genérico")


class Perro(Animal):

    def hablar(self):
        # Redefinimos el comportamiento heredado.
        print("Guau")


class Gato(Animal):

    def hablar(self):
        # Redefinimos el comportamiento heredado.
        print("Miau")


animales = [
    Perro(),
    Gato(),
    Perro()
]

for animal in animales:
    # La llamada es siempre la misma.
    animal.hablar()

    # Pero cada objeto responde según su tipo.
````

# 5.3. Herencia múltiple

Python permite que una clase herede de más de una clase base.

Por ejemplo:

```python
class C(B, A):
```

indica que `C` hereda de `B` y de `A`.

````python
class A:

    def __init__(self):
        print("Inicializador de A")

    def metodo_a(self):
        print("Este método se hereda de A")


class B:

    def __init__(self):
        print("Inicializador de B")

    def metodo_b(self):
        print("Este método se hereda de B")


class C(B, A):

    def metodo_c(self):
        print("Este método pertenece a C")


c = C()

# C hereda métodos tanto de A como de B.
c.metodo_a()
c.metodo_b()

# Y también dispone de sus propios métodos.
c.metodo_c()
````

## ¿Por qué se ejecuta el `__init__` de `B`?

Python necesita decidir en qué orden busca los métodos cuando intervienen varias clases.

Ese orden se denomina **MRO** (*Method Resolution Order*).

Podemos consultarlo mediante:

```python
C.__mro__
```

````python
print("Orden de resolución de métodos:")

for clase in C.__mro__:
    print(clase.__name__)
````

En este ejemplo el orden es:

```text
C
↓
B
↓
A
↓
object
```

Por eso, como `C` no define su propio `__init__`, Python encuentra primero `B.__init__`.

- todas las clases de Python heredan finalmente de `object`;
- el nombre de la clase base es `object`, en minúscula.

> La herencia múltiple es una herramienta potente, pero puede hacer más difícil comprender una jerarquía. En una introducción a POO basta con entender qué significa y cómo Python determina el orden de búsqueda.
