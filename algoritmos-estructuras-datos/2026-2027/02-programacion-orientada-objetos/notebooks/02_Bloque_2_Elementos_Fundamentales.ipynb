# 2.1. Atributos

Los **atributos** almacenan información asociada a una clase o a un objeto.

Por ejemplo, una galleta podría tener:

- sabor;
- color;
- peso;
- si tiene o no chocolate.

En Python conviene distinguir entre:

1. **atributos de clase**;
2. **atributos de instancia**.

## Atributo de clase

En el siguiente ejemplo `chocolate` se define dentro de la clase y fuera de cualquier método.

Por tanto, es un **atributo de clase**.

````python
class Galleta:
    # Atributo de clase.
    chocolate = False


galleta = Galleta()

# Como el objeto no tiene un atributo chocolate propio,
# Python utiliza el atributo definido en la clase.
print("¿Tiene chocolate?", galleta.chocolate)
````

## Modificar el atributo de clase

Podemos modificarlo directamente sobre la clase.

````python
class Galleta:
    chocolate = False


print("Valor inicial:", Galleta.chocolate)

# Modificamos el atributo en la propia clase.
Galleta.chocolate = True

galleta = Galleta()

print("Valor visto desde el objeto:", galleta.chocolate)
````

## Atributo de instancia

Un objeto también puede tener su propio valor.

Ese valor pertenece a esa instancia concreta.

````python
class Galleta:
    chocolate = False


galleta_1 = Galleta()
galleta_2 = Galleta()

# Creamos un atributo chocolate propio de galleta_1.
galleta_1.chocolate = True

print("galleta_1:", galleta_1.chocolate)
print("galleta_2:", galleta_2.chocolate)
print("Clase Galleta:", Galleta.chocolate)
````

## Atributos dinámicos

Python permite crear atributos directamente sobre un objeto.

Esto demuestra la flexibilidad del lenguaje, aunque en clases bien diseñadas normalmente preferiremos declarar de forma clara los atributos esperados, especialmente dentro de `__init__`.

````python
class Galleta:
    pass


galleta = Galleta()

# Creamos atributos "al vuelo".
galleta.sabor = "salado"
galleta.color = "marrón"

print(
    f"El sabor de esta galleta es {galleta.sabor} "
    f"y su color es {galleta.color}."
)
````

# 2.2. Métodos y `self`

Un **método** es una función definida dentro de una clase.

Representa normalmente un comportamiento del objeto.

Ejemplos:

- una galleta puede `chocolatear()`;
- un coche puede `acelerar()`;
- una cuenta puede `ingresar()` dinero.

En un método de instancia, el primer parámetro se escribe convencionalmente como `self`.

`self` representa **el objeto concreto sobre el que se está ejecutando el método**.

````python
class Galleta:
    chocolate = False

    def chocolatear(self):
        # Queremos modificar EL OBJETO.
        #
        # Por eso escribimos self.chocolate.
        # Si escribiéramos simplemente:
        #
        #     chocolate = True
        #
        # crearíamos una variable local dentro del método y
        # el atributo del objeto seguiría sin cambiar.
        self.chocolate = True


galleta = Galleta()

print("Antes:", galleta.chocolate)

galleta.chocolatear()

print("Después:", galleta.chocolate)
````

## ¿Qué hace realmente `self`?

Cuando escribimos:

```python
galleta.chocolatear()
```

podemos pensar, de forma simplificada, que Python está haciendo algo equivalente a:

```python
Galleta.chocolatear(galleta)
```

El objeto situado a la izquierda del punto termina siendo el valor de `self`.

````python
class Galleta:

    def saludar(self):
        print(f"Hola. Soy el objeto con id {id(self)}")


galleta = Galleta()

# Forma habitual.
galleta.saludar()

# Forma equivalente para entender self.
Galleta.saludar(galleta)
````

# 2.3. El constructor `__init__`

Cuando creamos un objeto solemos querer que nazca con unos valores iniciales.

Para ello utilizamos `__init__`.

Python lo ejecuta automáticamente cada vez que creamos una nueva instancia.

````python
class Galleta:

    def __init__(self):
        # Este código se ejecuta automáticamente
        # después de crear una nueva instancia.
        print("¡Soy una galleta recién creada!")


galleta = Galleta()
````

## `__init__` con parámetros

Lo normal es que distintos objetos nazcan con valores diferentes.

````python
class Galleta:

    def __init__(self, sabor, color):
        # sabor y color son parámetros.
        #
        # self.sabor y self.color son atributos
        # que quedarán almacenados dentro del objeto.
        self.sabor = sabor
        self.color = color

        print(
            f"Se acaba de crear una galleta "
            f"{self.color} y de sabor {self.sabor}."
        )


# El orden es Galleta(sabor, color).
galleta_1 = Galleta("amarga", "marrón")
galleta_2 = Galleta("dulce", "blanca")

print("\nDatos de galleta_1:")
print(galleta_1.sabor)
print(galleta_1.color)

print("\nDatos de galleta_2:")
print(galleta_2.sabor)
print(galleta_2.color)
````

### Una línea muy importante

Cuando escribimos:

```python
self.sabor = sabor
```

las dos apariciones de `sabor` no representan exactamente lo mismo:

- `sabor` → parámetro local recibido por el método;
- `self.sabor` → atributo que quedará guardado en el objeto.

# 2.4. Métodos especiales: `__str__` y `__len__`

Python dispone de métodos especiales cuyo nombre comienza y termina con dos guiones bajos.

A menudo se les denomina **dunder methods**.

Normalmente no necesitamos llamarlos directamente: Python los utiliza cuando realizamos determinadas operaciones.

## `__str__`

`__str__` devuelve una representación textual del objeto.

Python lo utiliza cuando hacemos:

```python
print(objeto)
```

o:

```python
str(objeto)
```

````python
class Galleta:

    def __init__(self, sabor, color):
        self.sabor = sabor
        self.color = color

    def __str__(self):
        # __str__ debe devolver una cadena.
        return f"Soy una galleta {self.color} y de sabor {self.sabor}."


galleta = Galleta("dulce", "blanca")

print(galleta)
print(str(galleta))

# También podríamos llamarlo directamente, pero no es la forma habitual.
print(galleta.__str__())
````

## `__len__`

Si una clase define `__len__`, podemos utilizar `len(objeto)`.

En este ejemplo interpretaremos la longitud de una canción como su duración en segundos.

````python
class Cancion:

    def __init__(self, autor, titulo, duracion):
        self.autor = autor
        self.titulo = titulo
        self.duracion = duracion

    def __len__(self):
        # __len__ debe devolver un entero mayor o igual que cero.
        return self.duracion

    def __str__(self):
        return f"{self.titulo} - {self.autor}"


cancion = Cancion(
    autor="Queen",
    titulo="Don't Stop Me Now",
    duracion=210
)

print(cancion)
print("Duración:", len(cancion), "segundos")
````

## Resumen del bloque

Debes poder distinguir:

- **atributo** → información o estado;
- **método** → comportamiento;
- `self` → objeto concreto sobre el que se ejecuta el método;
- `__init__` → inicializa cada nueva instancia;
- `__str__` → define cómo representar el objeto como texto;
- `__len__` → define qué significa `len(objeto)`.

### Error especialmente importante

Esto **no modifica** el atributo del objeto:

```python
def chocolatear(self):
    chocolate = True
```

Esto **sí modifica** el objeto:

```python
def chocolatear(self):
    self.chocolate = True
```
