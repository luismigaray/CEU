# Bloque 1. Introducción a la Programación Orientada a Objetos

## ¿Qué vamos a aprender?

En este primer bloque vamos a partir de un problema muy sencillo —gestionar clientes— para comparar dos formas de organizar un programa:

1. **Programación estructurada**: los datos se guardan por un lado y las funciones que trabajan con ellos se definen por otro.
2. **Programación orientada a objetos (POO)**: los datos y las operaciones que trabajan con esos datos se agrupan en objetos.

El objetivo no es memorizar todavía toda la sintaxis de las clases, sino entender **por qué la POO puede ayudarnos a organizar mejor un programa cuando empieza a crecer**.

> **Idea clave:** una clase puede entenderse como un **molde** y un objeto como una **instancia concreta** creada a partir de ese molde.

# 1.1. De programación estructurada a programación orientada a objetos

Empezamos utilizando programación estructurada.

Vamos a guardar varios clientes en una **lista**. Cada cliente será un **diccionario** con sus datos.

Después definiremos funciones para trabajar con esa lista.

Observa la separación:

- Los **datos** están en la variable `clientes`.
- Las **operaciones** están en funciones como `mostrar_cliente()` y `borrar_cliente()`.

````python
# ============================================================
# EJEMPLO 1: PROGRAMACIÓN ESTRUCTURADA
# ============================================================

# Una lista permite almacenar varios elementos.
# Cada elemento será un diccionario que representa a un cliente.
clientes = [
    {
        "nombre": "Hector",
        "apellidos": "Costa Guzman",
        "dni": "11111111A"
    },
    {
        "nombre": "Juan",
        "apellidos": "González Márquez",
        "dni": "22222222B"
    }
]


def mostrar_cliente(lista_clientes, dni):
    # Recorremos uno a uno los clientes almacenados en la lista.
    for cliente in lista_clientes:

        # Cada cliente es un diccionario.
        # Accedemos a su DNI mediante la clave "dni".
        if cliente["dni"] == dni:
            print(f'{cliente["nombre"]} {cliente["apellidos"]}')

            # return termina inmediatamente la función.
            # No es necesario seguir buscando.
            return

    # Si el bucle ha terminado sin encontrar el DNI, llegamos aquí.
    print("Cliente no encontrado")


def borrar_cliente(lista_clientes, dni):
    # enumerate() nos proporciona dos valores:
    # - posicion: índice del elemento.
    # - cliente: elemento almacenado en esa posición.
    for posicion, cliente in enumerate(lista_clientes):

        if cliente["dni"] == dni:
            # pop(posicion) elimina el elemento y, además,
            # devuelve el elemento que acabamos de eliminar.
            eliminado = lista_clientes.pop(posicion)

            print(
                f'{eliminado["nombre"]} {eliminado["apellidos"]} > BORRADO'
            )
            return

    print("Cliente no encontrado")


# ------------------------------------------------------------
# UTILIZACIÓN DEL PROGRAMA
# ------------------------------------------------------------

print("== LISTADO DE CLIENTES ==")
print(clientes)

print("\n== MOSTRAR CLIENTES POR DNI ==")
mostrar_cliente(clientes, "11111111A")  # Existe.
mostrar_cliente(clientes, "11111111Z")  # No existe.

print("\n== BORRAR CLIENTES POR DNI ==")
borrar_cliente(clientes, "22222222V")   # No existe.
borrar_cliente(clientes, "22222222B")   # Existe.

print("\n== LISTADO FINAL DE CLIENTES ==")
print(clientes)
````

### ¿Qué debemos observar?

El programa funciona, pero los datos y las operaciones están separados.

A medida que una aplicación crece podemos terminar teniendo:

- muchas estructuras de datos;
- muchas funciones;
- funciones que solo deberían trabajar con determinados datos;
- más dificultad para identificar qué operaciones pertenecen a cada concepto.

La POO propone **agrupar aquello que está relacionado**.

En nuestro ejemplo:

- un `Cliente` representa a una persona concreta;
- una `Empresa` administra una colección de clientes;
- las operaciones relacionadas con los clientes pueden convertirse en métodos del objeto `Empresa`.

## Primera versión orientada a objetos

Ahora resolvemos esencialmente el mismo problema utilizando dos clases:

- `Cliente`
- `Empresa`

No es necesario comprender todavía todos los detalles de `__init__`, `self` o `__str__`. Los veremos en el siguiente bloque.

Por ahora, céntrate en **cómo cambia la forma de utilizar el programa**.

````python
# ============================================================
# EJEMPLO 2: PRIMERA VERSIÓN ORIENTADA A OBJETOS
# ============================================================

class Cliente:

    def __init__(self, dni, nombre, apellidos):
        # Guardamos dentro del objeto los datos del cliente.
        self.dni = dni
        self.nombre = nombre
        self.apellidos = apellidos

    def __str__(self):
        # Indicamos cómo queremos mostrar un cliente como texto.
        return f"{self.nombre} {self.apellidos}"


class Empresa:

    def __init__(self, clientes=None):
        # IMPORTANTE:
        # Evitamos escribir clientes=[] como argumento por defecto.
        #
        # Las listas son mutables. Utilizar una lista como valor
        # por defecto puede provocar que distintas instancias
        # terminen compartiendo la misma lista.
        if clientes is None:
            self.clientes = []
        else:
            # Creamos una copia de la lista recibida.
            self.clientes = list(clientes)

    def listar_clientes(self):
        # Comprobamos primero si la lista está vacía.
        if not self.clientes:
            print("No hay clientes.")
            return

        # Cada elemento de la lista es ahora un OBJETO Cliente.
        for cliente in self.clientes:
            print(cliente)

    def mostrar_cliente(self, dni):
        # Recorremos los objetos Cliente almacenados en la empresa.
        for cliente in self.clientes:

            # Ya no utilizamos cliente["dni"].
            # Ahora accedemos al atributo del objeto: cliente.dni
            if cliente.dni == dni:
                print(cliente)
                return

        print("Cliente no encontrado")

    def borrar_cliente(self, dni):
        for posicion, cliente in enumerate(self.clientes):
            if cliente.dni == dni:
                eliminado = self.clientes.pop(posicion)
                print(f"{eliminado} > BORRADO")
                return

        print("Cliente no encontrado")


# ------------------------------------------------------------
# CREAMOS OBJETOS Cliente
# ------------------------------------------------------------

hector = Cliente(
    dni="11111111A",
    nombre="Hector",
    apellidos="Costa Guzman"
)

juan = Cliente(
    dni="22222222B",
    nombre="Juan",
    apellidos="González Márquez"
)


# ------------------------------------------------------------
# CREAMOS UN OBJETO Empresa
# ------------------------------------------------------------

empresa = Empresa(clientes=[hector, juan])


# ------------------------------------------------------------
# UTILIZAMOS EL OBJETO
# ------------------------------------------------------------

print("== LISTADO DE CLIENTES ==")
empresa.listar_clientes()

print("\n== MOSTRAR CLIENTES POR DNI ==")
empresa.mostrar_cliente("11111111A")
empresa.mostrar_cliente("11111111Z")

print("\n== BORRAR CLIENTES POR DNI ==")
empresa.borrar_cliente("22222222V")
empresa.borrar_cliente("22222222B")

print("\n== LISTADO FINAL DE CLIENTES ==")
empresa.listar_clientes()
````

### Comparación rápida

| Programación estructurada | Programación orientada a objetos |
|---|---|
| Los clientes eran diccionarios | Los clientes son objetos `Cliente` |
| La lista estaba en una variable independiente | La lista pertenece al objeto `Empresa` |
| `mostrar_cliente(clientes, dni)` | `empresa.mostrar_cliente(dni)` |
| `borrar_cliente(clientes, dni)` | `empresa.borrar_cliente(dni)` |

Fíjate especialmente en estas dos formas:

```python
mostrar_cliente(clientes, "11111111A")
```

y:

```python
empresa.mostrar_cliente("11111111A")
```

En la segunda podemos leer mentalmente:

> «Objeto `empresa`: ejecuta tu método `mostrar_cliente`».

Esta forma de pensar es fundamental en POO.

# 1.2. Clases, objetos e instancias

Vamos ahora a reducir el problema al mínimo.

Una **clase** es una definición o molde.

Un **objeto** es una entidad concreta creada a partir de una clase.

La palabra **instancia** se utiliza para indicar que un objeto concreto ha sido creado a partir de una determinada clase.

Podemos imaginar:

```text
Clase Galleta
      │
      ├── objeto una_galleta
      └── objeto otra_galleta
```

````python
# Definimos una clase extremadamente sencilla.
#
# La clase todavía no contiene atributos ni métodos.
# pass significa: "no hacer nada".
# Nos permite dejar el cuerpo de la clase vacío sin producir un error.

class Galleta:
    pass
````

````python
# Creamos dos objetos distintos a partir de la misma clase.

una_galleta = Galleta()
otra_galleta = Galleta()

# Los dos objetos son instancias de Galleta,
# pero son objetos diferentes.
print(una_galleta)
print(otra_galleta)

# id() devuelve un identificador del objeto durante su vida.
# Los valores serán diferentes porque son dos objetos distintos.
print("\nIdentificadores:")
print(id(una_galleta))
print(id(otra_galleta))
````

## ¿Qué guarda la variable?

Cuando escribimos:

```python
una_galleta = Galleta()
```

la variable `una_galleta` nos permite **referenciar** el objeto que acabamos de crear.

A partir de esa referencia podremos acceder a los atributos y métodos del objeto mediante la notación punto:

```python
una_galleta.atributo
una_galleta.metodo()
```

````python
# Podemos consultar la clase de un objeto de varias formas.

print("Nombre de la clase:", Galleta.__name__)
print("Tipo del objeto:", type(una_galleta))
print("Nombre del tipo:", type(una_galleta).__name__)
print("Clase del objeto:", una_galleta.__class__)
print("Nombre de la clase:", una_galleta.__class__.__name__)

# isinstance() responde a la pregunta:
# "¿este objeto es una instancia de esta clase?"
print("\n¿una_galleta es una instancia de Galleta?")
print(isinstance(una_galleta, Galleta))
````
