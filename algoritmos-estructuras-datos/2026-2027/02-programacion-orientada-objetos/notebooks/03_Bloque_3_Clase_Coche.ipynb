# 3.1. Diseño y definición de la clase `Coche`

Cada coche tendrá información propia:

- marca;
- color;
- combustible;
- velocidad máxima;
- velocidad actual;
- estado.

Y ofrecerá comportamientos:

- encender;
- apagar;
- acelerar;
- frenar;
- cambiar de color.

Esta separación entre **estado** y **comportamiento** es una de las ideas centrales de la POO.

````python
class Coche:

    def __init__(
        self,
        marca,
        color,
        combustible,
        velocidad_maxima,
        velocidad_actual=0,
        estado="apagado"
    ):
        # --------------------------------------------------------
        # ESTADO DEL OBJETO
        # --------------------------------------------------------

        self.marca = marca
        self.color = color
        self.combustible = combustible
        self.velocidad_maxima = velocidad_maxima
        self.velocidad_actual = velocidad_actual
        self.estado = estado

    def encender(self):
        # Si ya estaba encendido, no hacemos nada.
        if self.estado == "encendido":
            print("El coche ya estaba encendido.")
            return

        self.estado = "encendido"
        print("Coche encendido.")

    def apagar(self):
        # Introducimos una regla sencilla:
        # no permitimos apagar un coche que está en movimiento.
        if self.velocidad_actual > 0:
            print("No se puede apagar el coche mientras está en movimiento.")
            return

        self.estado = "apagado"
        print("Coche apagado.")

    def acelerar(self, incremento=10):
        # Solo podemos acelerar si el coche está encendido.
        if self.estado != "encendido":
            print("El coche está apagado. No se puede acelerar.")
            return

        if incremento <= 0:
            print("El incremento debe ser positivo.")
            return

        nueva_velocidad = self.velocidad_actual + incremento

        # min() evita superar la velocidad máxima.
        self.velocidad_actual = min(
            nueva_velocidad,
            self.velocidad_maxima
        )

    def frenar(self, decremento=10):
        # Solo frenamos si el coche está encendido.
        if self.estado != "encendido":
            print("El coche está apagado. No se puede frenar.")
            return

        if decremento <= 0:
            print("El decremento debe ser positivo.")
            return

        nueva_velocidad = self.velocidad_actual - decremento

        # max() evita velocidades negativas.
        self.velocidad_actual = max(nueva_velocidad, 0)

    def cambiar_color(self, nuevo_color):
        # self.color pertenece al objeto concreto.
        self.color = nuevo_color

    def __str__(self):
        return (
            f"Coche("
            f"marca={self.marca}, "
            f"color={self.color}, "
            f"combustible={self.combustible}, "
            f"velocidad={self.velocidad_actual}/{self.velocidad_maxima} km/h, "
            f"estado={self.estado}"
            f")"
        )
````

### Corrección importante del código original

En el material original aparecía:

```python
def cambiarColor(coche, color):
    coche.color = color
```

Python lo acepta, pero la convención es utilizar `self` como primer parámetro:

```python
def cambiar_color(self, nuevo_color):
    self.color = nuevo_color
```

También utilizamos `snake_case`, que es la convención habitual de Python para nombres de métodos.

# 3.2. Creación de instancias y utilización de métodos

Creamos tres objetos.

Todos pertenecen a la clase `Coche`, pero cada uno conserva su propio estado.

````python
coche_1 = Coche(
    marca="Ford",
    color="Rojo",
    combustible="Gasolina",
    velocidad_maxima=200
)

coche_2 = Coche(
    marca="Seat",
    color="Azul",
    combustible="Gasolina",
    velocidad_maxima=180
)

coche_3 = Coche(
    marca="Audi",
    color="Verde",
    combustible="Gasolina",
    velocidad_maxima=220
)

print(coche_1)
print(coche_2)
print(coche_3)
````

## Invocación de métodos

Probemos primero a acelerar con el coche apagado.

````python
print("Estado inicial:")
print(coche_1)

print("\nIntentamos acelerar:")
coche_1.acelerar()

print("\nEncendemos el coche:")
coche_1.encender()

print("\nAceleramos 10 km/h:")
coche_1.acelerar()

print("\nAceleramos otros 25 km/h:")
coche_1.acelerar(25)

print("\nEstado final:")
print(coche_1)
````

## Frenar y cambiar el color

````python
coche_1.frenar(15)
print(coche_1)

coche_1.cambiar_color("Amarillo")
print(coche_1)
````

## Error muy frecuente: olvidar los paréntesis

Estas expresiones no hacen lo mismo:

```python
coche_2.acelerar
```

```python
coche_2.acelerar()
```

- La primera **referencia el método**.
- La segunda **ejecuta el método**.

En el código original había una llamada sin paréntesis. El programa no generaba un error, pero el método tampoco se ejecutaba.

````python
coche_2.encender()

print("Esto es la referencia al método:")
print(coche_2.acelerar)

print("\nAhora sí ejecutamos el método:")
coche_2.acelerar()

print(coche_2)
````

## Cada objeto conserva su propio estado

Modificar `coche_1` no modifica `coche_2` ni `coche_3`.

````python
print("coche_1:", coche_1)
print("coche_2:", coche_2)
print("coche_3:", coche_3)
````
