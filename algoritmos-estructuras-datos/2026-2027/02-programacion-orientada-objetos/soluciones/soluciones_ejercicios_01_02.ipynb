## Ejercicio del bloque 1. Crear una clase y varios objetos

Vamos a crear una clase sencilla llamada `Mascota`.

### Enunciado

Crea una clase `Mascota` que permita representar mascotas.

Cada mascota debe guardar la siguiente información:

- `nombre`
- `tipo` (por ejemplo: `"Perro"`, `"Gato"`, `"Conejo"`, etc.)
- `edad`

Después:

1. Crea **tres objetos** de la clase `Mascota`.
2. Utiliza nombres y datos diferentes para cada uno.
3. Muestra por pantalla el `nombre`, el `tipo` y la `edad` de cada mascota.
4. Comprueba mediante `isinstance()` que uno de los objetos creados es una instancia de `Mascota`.

### Ejemplo orientativo de salida

```text
Luna - Gato - 3 años
Toby - Perro - 5 años
Nube - Conejo - 2 años

¿Luna es una Mascota? True
```

> **Objetivo del ejercicio:** practicar la creación de una clase, la creación de objetos y el acceso a los atributos de cada instancia.

### Solución propuesta

````python
# ============================================================
# SOLUCIÓN PROPUESTA
# ============================================================

class Mascota:

    def __init__(self, nombre, tipo, edad):
        # Guardamos dentro de cada objeto los datos de la mascota.
        self.nombre = nombre
        self.tipo = tipo
        self.edad = edad


# ------------------------------------------------------------
# CREAMOS TRES OBJETOS DIFERENTES
# ------------------------------------------------------------

mascota_1 = Mascota("Luna", "Gato", 3)
mascota_2 = Mascota("Toby", "Perro", 5)
mascota_3 = Mascota("Nube", "Conejo", 2)


# ------------------------------------------------------------
# MOSTRAMOS LOS DATOS DE CADA OBJETO
# ------------------------------------------------------------

print(
    f"{mascota_1.nombre} - "
    f"{mascota_1.tipo} - "
    f"{mascota_1.edad} años"
)

print(
    f"{mascota_2.nombre} - "
    f"{mascota_2.tipo} - "
    f"{mascota_2.edad} años"
)

print(
    f"{mascota_3.nombre} - "
    f"{mascota_3.tipo} - "
    f"{mascota_3.edad} años"
)


# ------------------------------------------------------------
# COMPROBAMOS QUE EL OBJETO PERTENECE A LA CLASE Mascota
# ------------------------------------------------------------

print(
    "\n¿Luna es una Mascota?",
    isinstance(mascota_1, Mascota)
)
````

### Explicación de la solución

La clase `Mascota` funciona como un molde.

Cada vez que hacemos:

```python
Mascota("Luna", "Gato", 3)
```

se crea un objeto nuevo con sus propios valores.

Por eso `mascota_1`, `mascota_2` y `mascota_3` pertenecen a la misma clase, pero contienen datos diferentes.

Además:

```python
isinstance(mascota_1, Mascota)
```

comprueba si `mascota_1` es una instancia de la clase `Mascota`.

## Ejercicio del bloque 2. Clase `Libro`

### Enunciado

Crea una clase llamada `Libro`.

Cada libro debe tener los siguientes atributos:

- `titulo`
- `autor`
- `paginas`

La clase debe incluir:

1. Un constructor `__init__` que inicialice los tres atributos.
2. Un método llamado `descripcion()` que muestre el título y el autor.
3. Un método especial `__str__()` que devuelva un texto con el formato:

```text
El Quijote - Miguel de Cervantes
```

4. Un método especial `__len__()` que devuelva el número de páginas del libro.

Después:

5. Crea un objeto `Libro`.
6. Ejecuta `descripcion()`.
7. Muestra el objeto con `print()`.
8. Utiliza `len()` para mostrar su número de páginas.

> **Objetivo del ejercicio:** practicar atributos, `self`, métodos, `__init__`, `__str__` y `__len__`.

### Solución propuesta

````python
# ============================================================
# SOLUCIÓN PROPUESTA
# ============================================================

class Libro:

    def __init__(self, titulo, autor, paginas):
        # Los parámetros recibidos se almacenan
        # como atributos del objeto.
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descripcion(self):
        # self permite acceder a los datos
        # del objeto concreto que ejecuta el método.
        print(
            f'Título: "{self.titulo}" | '
            f'Autor: {self.autor}'
        )

    def __str__(self):
        # Python utilizará este método cuando hagamos print(libro).
        return f"{self.titulo} - {self.autor}"

    def __len__(self):
        # len(libro) devolverá el número de páginas.
        return self.paginas


# ------------------------------------------------------------
# CREAMOS UN OBJETO
# ------------------------------------------------------------

libro = Libro(
    titulo="El Quijote",
    autor="Miguel de Cervantes",
    paginas=863
)


# ------------------------------------------------------------
# UTILIZAMOS SUS MÉTODOS
# ------------------------------------------------------------

libro.descripcion()

print("Representación del objeto:")
print(libro)

print("Número de páginas:")
print(len(libro))
````

### Explicación de la solución

En este ejercicio aparecen juntos los principales elementos del bloque:

```text
Libro
├── atributos
│   ├── titulo
│   ├── autor
│   └── paginas
│
└── métodos
    ├── descripcion()
    ├── __str__()
    └── __len__()
```

`descripcion()` es un método normal de instancia.

`__str__()` permite decidir qué muestra:

```python
print(libro)
```

y `__len__()` permite utilizar:

```python
len(libro)
```
