## Ejercicio 1. El termómetro inteligente

El programa convertirá una temperatura de grados Celsius a Fahrenheit y
la clasificará como frío, temperatura moderada o calor.

Separaremos el cálculo de la lectura por teclado. De este modo podemos
probar la lógica sin tener que escribir datos cada vez.


### 1. Conversión y clasificación

La fórmula es F = C × 9/5 + 32. Los límites se comprueban en orden:
primero menos de 10 grados, después menos de 25 y, finalmente, el resto.


```python
def clasificar_temperatura(celsius):
    if celsius < 10:
        return "frío"
    elif celsius < 25:
        return "temperatura moderada"
    else:
        return "calor"


def convertir_temperatura(celsius):
    fahrenheit = celsius * 9 / 5 + 32
    clasificacion = clasificar_temperatura(celsius)
    return fahrenheit, clasificacion
```

```python
celsius = 20
fahrenheit, clasificacion = convertir_temperatura(celsius)

print(f"{celsius:.1f} ºC equivalen a {fahrenheit:.1f} ºF.")
print(f"Clasificación: {clasificacion}.")
```

### 2. Lectura segura

La conversión con float puede lanzar ValueError. La capturamos para
mostrar un mensaje comprensible en lugar de terminar el programa.


```python
def leer_temperatura():
    try:
        return float(input("Introduce la temperatura en ºC: "))
    except ValueError:
        print("Error: debes introducir un número.")
        return None


def ejecutar_termometro():
    celsius = leer_temperatura()
    if celsius is None:
        return

    fahrenheit, clasificacion = convertir_temperatura(celsius)
    print(f"{celsius:.1f} ºC equivalen a {fahrenheit:.1f} ºF.")
    print(f"Clasificación: {clasificacion}.")


# Descomenta la línea siguiente para utilizar el programa interactivo.
# ejecutar_termometro()
```

### 3. Comprobaciones


```python
assert convertir_temperatura(20) == (68.0, "temperatura moderada")
assert clasificar_temperatura(9.9) == "frío"
assert clasificar_temperatura(10) == "temperatura moderada"
assert clasificar_temperatura(25) == "calor"

print("Comprobaciones superadas.")
```

## Ejercicio 2. La semana de temperaturas

Recorreremos una lista de siete temperaturas. Durante el recorrido
acumularemos la suma para calcular la media. Después construiremos una
nueva lista con los valores superiores a esa media.


```python
temperaturas = [18, 21, 19, 24, 27, 25, 20]
```

### 1. Función de análisis

La lista no puede estar vacía porque no se podría calcular su media.


```python
def analizar_temperaturas(valores):
    if not valores:
        raise ValueError("La lista de temperaturas no puede estar vacía.")

    suma = 0
    for temperatura in valores:
        suma += temperatura

    media = suma / len(valores)
    superiores = []

    for temperatura in valores:
        if temperatura > media:
            superiores.append(temperatura)

    return {
        "maxima": max(valores),
        "minima": min(valores),
        "media": media,
        "dias_superiores": len(superiores),
        "temperaturas_superiores": superiores,
    }
```

```python
resultado = analizar_temperaturas(temperaturas)

print(f"Temperatura máxima: {resultado['maxima']} ºC")
print(f"Temperatura mínima: {resultado['minima']} ºC")
print(f"Temperatura media: {resultado['media']:.2f} ºC")
print(f"Días por encima de la media: {resultado['dias_superiores']}")
print(f"Temperaturas superiores: {resultado['temperaturas_superiores']}")
```

### 2. Comprobaciones


```python
assert resultado["maxima"] == 27
assert resultado["minima"] == 18
assert resultado["media"] == 22
assert resultado["temperaturas_superiores"] == [24, 27, 25]
assert resultado["dias_superiores"] == 3

# Segundo conjunto de datos para comprobar que la función es general.
otro_resultado = analizar_temperaturas([5, 5, 10])
assert otro_resultado["maxima"] == 10
assert otro_resultado["minima"] == 5
assert abs(otro_resultado["media"] - 20 / 3) < 1e-9
assert otro_resultado["temperaturas_superiores"] == [10]
assert otro_resultado["dias_superiores"] == 1

print("Comprobaciones superadas con dos conjuntos de datos.")
```

## Ejercicio 3. ¿Cuánta electricidad hemos gastado?

Calcularemos el consumo total, la media diaria, el coste y los días con
más de 10 kWh. Usamos enumerate con inicio en 1 para numerar los días
de la misma forma que en el enunciado.


```python
consumos = [8.2, 7.5, 9.1, 12.0, 6.8, 10.4, 11.2]
precio_kwh = 0.18
```

```python
def analizar_consumos(valores, precio):
    if not valores:
        raise ValueError("La lista de consumos no puede estar vacía.")

    total = 0
    dias_consumo_alto = []

    for dia, consumo in enumerate(valores, start=1):
        total += consumo
        if consumo > 10:
            dias_consumo_alto.append(dia)

    media = total / len(valores)
    coste = total * precio

    return total, media, coste, dias_consumo_alto
```

```python
total, media, coste, dias_altos = analizar_consumos(consumos, precio_kwh)

print(f"Consumo total: {total:.2f} kWh")
print(f"Consumo medio diario: {media:.2f} kWh")
print(f"Coste total: {coste:.2f} €")
print(f"Días con consumo alto: {dias_altos}")
```

### Comprobaciones


```python
assert abs(total - 65.2) < 1e-9
assert abs(media - 65.2 / 7) < 1e-9
assert abs(coste - 11.736) < 1e-9
assert dias_altos == [4, 6, 7]

# Segundo caso: un consumo de exactamente 10 kWh no cuenta como alto.
otro_total, otra_media, otro_coste, otros_dias = analizar_consumos([10, 11], 0.20)
assert otro_total == 21
assert otra_media == 10.5
assert abs(otro_coste - 4.2) < 1e-9
assert otros_dias == [2]

print("Comprobaciones superadas con dos conjuntos de datos.")
```

## Ejercicio 4. Una función que analiza cualquier lista de datos

La función recibe una lista y devuelve tres resultados: media, mínimo y
máximo. Si la lista está vacía lanzamos ValueError, porque no es posible
dividir entre cero ni obtener sus extremos.


```python
def analizar_datos(datos):
    if not datos:
        raise ValueError("La lista de datos no puede estar vacía.")

    media = sum(datos) / len(datos)
    minimo = min(datos)
    maximo = max(datos)
    return media, minimo, maximo
```

### Ejemplos de uso


```python
ejemplo_1 = analizar_datos([2, 4, 6])
ejemplo_2 = analizar_datos([10, -2, 5, 7])

print(f"Primer ejemplo: {ejemplo_1}")
print(f"Segundo ejemplo: {ejemplo_2}")
```

### Comprobaciones


```python
assert analizar_datos([2, 4, 6]) == (4, 2, 6)
assert analizar_datos([5]) == (5, 5, 5)

try:
    analizar_datos([])
    raise AssertionError("La lista vacía debería producir ValueError.")
except ValueError:
    pass

print("Comprobaciones superadas.")
```

## Ejercicio 5. Calculadora segura

Separamos la lectura de números, el cálculo y la interacción principal.
Así cada parte tiene una responsabilidad sencilla y puede probarse por
separado.


### 1. Leer un número hasta que sea válido


```python
def leer_numero(mensaje):
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print("Valor no válido. Introduce un número.")
```

### 2. Realizar la operación

La división entre cero y una operación desconocida se indican mediante
excepciones concretas. El programa principal las transformará en
mensajes claros.


```python
def calcular(primero, segundo, operacion):
    if operacion == "+":
        return primero + segundo
    elif operacion == "-":
        return primero - segundo
    elif operacion == "*":
        return primero * segundo
    elif operacion == "/":
        if segundo == 0:
            raise ZeroDivisionError("No se puede dividir entre cero.")
        return primero / segundo
    else:
        raise ValueError("La operación debe ser +, -, * o /.")
```

```python
def ejecutar_calculadora():
    print("Operaciones disponibles: +, -, * y /")
    operacion = input("Elige una operación: ").strip()
    primero = leer_numero("Primer número: ")
    segundo = leer_numero("Segundo número: ")

    try:
        resultado = calcular(primero, segundo, operacion)
        print(f"Resultado: {resultado}")
    except (ValueError, ZeroDivisionError) as error:
        print(f"Error: {error}")


# Descomenta la línea siguiente para utilizar la calculadora.
# ejecutar_calculadora()
```

### 3. Comprobaciones


```python
assert calcular(3, 2, "+") == 5
assert calcular(3, 2, "-") == 1
assert calcular(3, 2, "*") == 6
assert calcular(8, 2, "/") == 4

try:
    calcular(5, 0, "/")
    raise AssertionError("La división entre cero debería fallar.")
except ZeroDivisionError:
    pass

print("Comprobaciones superadas.")
```
