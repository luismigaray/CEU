## Ejercicio 6. Ficha de un alumno

Leeremos una nota válida entre 0 y 10, la clasificaremos y guardaremos
el nombre y la nota en un diccionario.


### 1. Clasificar y crear la ficha


```python
def clasificar_nota(nota):
    if not 0 <= nota <= 10:
        raise ValueError("La nota debe estar entre 0 y 10.")

    if nota < 5:
        return "Suspenso"
    elif nota < 7:
        return "Aprobado"
    elif nota < 9:
        return "Notable"
    else:
        return "Sobresaliente"


def crear_ficha(nombre, nota):
    clasificar_nota(nota)
    return {"nombre": nombre, "nota": nota}
```

### 2. Pedir la nota hasta que sea correcta


```python
def leer_nota():
    while True:
        try:
            nota = float(input("Introduce una nota entre 0 y 10: "))
            if 0 <= nota <= 10:
                return nota
            print("La nota debe estar entre 0 y 10.")
        except ValueError:
            print("Debes introducir un número.")


def ejecutar_ficha():
    nombre = input("Nombre del alumno: ").strip()
    nota = leer_nota()
    alumno = crear_ficha(nombre, nota)
    clasificacion = clasificar_nota(alumno["nota"])
    print(f"{alumno['nombre']} ha obtenido un {alumno['nota']} → {clasificacion}")


# Descomenta la línea siguiente para introducir tus propios datos.
# ejecutar_ficha()
```

### 3. Ejemplo y comprobaciones


```python
alumno = crear_ficha("Ana", 8.2)
print(f"{alumno['nombre']} ha obtenido un {alumno['nota']} → "
      f"{clasificar_nota(alumno['nota'])}")

assert clasificar_nota(4.9) == "Suspenso"
assert clasificar_nota(5) == "Aprobado"
assert clasificar_nota(7) == "Notable"
assert clasificar_nota(9) == "Sobresaliente"

print("Comprobaciones superadas.")
```

## Ejercicio 7. Analizar un grupo de alumnos

El grupo se representa mediante una lista de diccionarios. En un único
recorrido sumaremos las notas, guardaremos los aprobados y localizaremos
al alumno con la nota más alta.


```python
alumnos = [
    {"nombre": "Ana", "nota": 8.5},
    {"nombre": "Luis", "nota": 4.2},
    {"nombre": "María", "nota": 9.1},
    {"nombre": "Pablo", "nota": 6.0},
    {"nombre": "Elena", "nota": 5.5},
]
```

```python
def analizar_grupo(grupo):
    if not grupo:
        raise ValueError("El grupo no puede estar vacío.")

    suma = 0
    aprobados = []
    mejor_alumno = grupo[0]

    for alumno in grupo:
        suma += alumno["nota"]

        if alumno["nota"] >= 5:
            aprobados.append(alumno["nombre"])

        if alumno["nota"] > mejor_alumno["nota"]:
            mejor_alumno = alumno

    media = suma / len(grupo)
    return media, aprobados, mejor_alumno
```

```python
for alumno in alumnos:
    print(f"{alumno['nombre']}: {alumno['nota']}")

media, aprobados, mejor = analizar_grupo(alumnos)
print(f"Nota media: {media:.2f}")
print(f"Alumnos aprobados: {aprobados}")
print(f"Mejor nota: {mejor['nombre']} con {mejor['nota']}")
```

### Comprobaciones


```python
assert abs(media - 6.66) < 1e-9
assert aprobados == ["Ana", "María", "Pablo", "Elena"]
assert mejor == {"nombre": "María", "nota": 9.1}

# Segundo grupo para verificar que el análisis no depende de los datos iniciales.
grupo_adicional = [
    {"nombre": "Lola", "nota": 4},
    {"nombre": "Diego", "nota": 5},
]
otra_media, otros_aprobados, otro_mejor = analizar_grupo(grupo_adicional)
assert otra_media == 4.5
assert otros_aprobados == ["Diego"]
assert otro_mejor == {"nombre": "Diego", "nota": 5}

print("Comprobaciones superadas con dos grupos.")
```

## Ejercicio 8. Buscador de productos y precios

El diccionario relaciona cada producto con su precio. Normalizamos la
consulta con lower para que las mayúsculas no afecten a la búsqueda.
El método get evita que aparezca KeyError cuando el producto no existe.


```python
precios = {
    "pan": 1.20,
    "leche": 1.05,
    "café": 3.50,
    "arroz": 1.80,
    "huevos": 2.40,
}
```

```python
def buscar_precio(nombre, catalogo):
    producto = nombre.strip().lower()
    return catalogo.get(producto)


def mostrar_productos(catalogo):
    print("Productos disponibles:")
    for producto in catalogo:
        print(f"- {producto}")
```

```python
def ejecutar_buscador():
    consulta = input("Producto que deseas consultar: ")
    precio = buscar_precio(consulta, precios)

    if precio is None:
        print("El producto no está disponible.")
    else:
        print(f"Precio: {precio:.2f} €")

    mostrar_productos(precios)


# Descomenta la línea siguiente para realizar una consulta.
# ejecutar_buscador()
```

### Ejemplo y comprobaciones


```python
consulta = "CAFÉ"
precio = buscar_precio(consulta, precios)
print(f"{consulta} cuesta {precio:.2f} €")

assert buscar_precio("PAN", precios) == 1.20
assert buscar_precio("CAFÉ", precios) == 3.50
assert buscar_precio("  leche  ", precios) == 1.05
assert buscar_precio("tomates", precios) is None

print("Comprobaciones superadas.")
```

## Ejercicio 9. Funciones para gestionar alumnos

Cada operación se coloca en una función independiente. Esto evita
repetir código y facilita comprobar cada comportamiento por separado.


```python
alumnos = [
    {"nombre": "Ana", "nota": 8.5},
    {"nombre": "Luis", "nota": 4.2},
    {"nombre": "María", "nota": 9.1},
    {"nombre": "Pablo", "nota": 6.0},
    {"nombre": "Elena", "nota": 5.5},
]
```

```python
def es_aprobado(nota):
    return nota >= 5


def calcular_media(grupo):
    if not grupo:
        raise ValueError("No se puede calcular la media de un grupo vacío.")

    suma = 0
    for alumno in grupo:
        suma += alumno["nota"]

    return suma / len(grupo)


def buscar_alumno(nombre, grupo):
    nombre_normalizado = nombre.strip().lower()

    for alumno in grupo:
        if alumno["nombre"].lower() == nombre_normalizado:
            return alumno

    return None
```

### Ejemplos de uso


```python
print(f"¿Está aprobado un 6? {es_aprobado(6)}")
print(f"Media del grupo: {calcular_media(alumnos):.2f}")
print(f"Búsqueda de ana: {buscar_alumno('ana', alumnos)}")
```

### Comprobaciones


```python
assert es_aprobado(5) is True
assert es_aprobado(4.9) is False
assert abs(calcular_media(alumnos) - 6.66) < 1e-9
assert buscar_alumno("ana", alumnos)["nombre"] == "Ana"
assert buscar_alumno("PEDRO", alumnos) is None

print("Comprobaciones superadas.")
```

## Ejercicio 10. Mini sistema de gestión de alumnos

Reunimos listas, diccionarios, funciones, bucles y validación en un
programa con menú. Las funciones que modifican o consultan la lista se
mantienen separadas de la interacción por teclado.

En los identificadores usamos anadir en lugar de añadir para evitar
posibles problemas de teclado o codificación al reutilizar el archivo.


### 1. Operaciones sobre la lista


```python
def validar_nota(nota):
    if not 0 <= nota <= 10:
        raise ValueError("La nota debe estar entre 0 y 10.")


def anadir_alumno(grupo, nombre, nota):
    validar_nota(nota)
    grupo.append({"nombre": nombre, "nota": nota})


def obtener_aprobados(grupo):
    aprobados = []
    for alumno in grupo:
        if alumno["nota"] >= 5:
            aprobados.append(alumno)
    return aprobados


def calcular_media(grupo):
    if not grupo:
        raise ValueError("Todavía no hay alumnos.")

    suma = 0
    for alumno in grupo:
        suma += alumno["nota"]
    return suma / len(grupo)


def mostrar_alumnos(grupo, mensaje_vacio="Todavía no hay alumnos."):
    if not grupo:
        print(mensaje_vacio)
        return

    for alumno in grupo:
        print(f"{alumno['nombre']}: {alumno['nota']}")
```

### 2. Lectura segura de la nota


```python
def leer_nota():
    while True:
        try:
            nota = float(input("Nota entre 0 y 10: "))
            validar_nota(nota)
            return nota
        except ValueError as error:
            print(f"Error: {error}")
```

### 3. Menú principal

El bucle continúa hasta seleccionar la opción 5. La llamada final está
comentada para que el notebook se pueda ejecutar completo sin detenerse
esperando datos.


```python
def ejecutar_menu():
    alumnos = []

    while True:
        print("\n1) Añadir alumno")
        print("2) Mostrar alumnos")
        print("3) Mostrar aprobados")
        print("4) Calcular media")
        print("5) Salir")

        opcion = input("Elige una opción: ").strip()

        if opcion == "1":
            nombre = input("Nombre: ").strip()
            nota = leer_nota()
            anadir_alumno(alumnos, nombre, nota)
        elif opcion == "2":
            mostrar_alumnos(alumnos)
        elif opcion == "3":
            mostrar_alumnos(
                obtener_aprobados(alumnos),
                "No hay alumnos aprobados.",
            )
        elif opcion == "4":
            try:
                print(f"Media: {calcular_media(alumnos):.2f}")
            except ValueError as error:
                print(error)
        elif opcion == "5":
            print("Programa finalizado.")
            break
        else:
            print("Opción no válida.")


# Descomenta la línea siguiente para iniciar el programa.
# ejecutar_menu()
```

### 4. Comprobación sin introducir datos


```python
grupo_prueba = []
anadir_alumno(grupo_prueba, "Ana", 8)
anadir_alumno(grupo_prueba, "Luis", 4)

assert obtener_aprobados(grupo_prueba) == [
    {"nombre": "Ana", "nota": 8}
]
assert calcular_media(grupo_prueba) == 6

# Segundo grupo: permite comprobar también el caso sin aprobados.
grupo_suspensos = []
anadir_alumno(grupo_suspensos, "Marta", 3)
anadir_alumno(grupo_suspensos, "Jorge", 4)
assert obtener_aprobados(grupo_suspensos) == []
assert calcular_media(grupo_suspensos) == 3.5

print("Alumnos:")
mostrar_alumnos(grupo_prueba)
print("Aprobados:")
mostrar_alumnos(obtener_aprobados(grupo_prueba), "No hay alumnos aprobados.")
print("Aprobados del segundo grupo:")
mostrar_alumnos(obtener_aprobados(grupo_suspensos), "No hay alumnos aprobados.")
print("Comprobaciones superadas con dos grupos.")
```
